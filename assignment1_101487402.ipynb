```python
"""
Author: Shifa Karnelia
Assignment: #1
"""


#q(b): create variable and comment datatype 

gym_member = "Alex Alliton" #Datatype: Str (string)
preferred_weight_kg = 20.5 #Datatype: float
highest_reps = 25 #Datatype: int
membership_active = True #Datatype: bool (boolean)

#q(c)
#dictionary : where key = Strings representing the names of friends 
#             value = Tuples containing the minutes spent on three different workout activities 
workout_stats = {"Alex":(30,45,25),
                 "Jamie":(20,40,15), 
                 "Taylor":(40,40,45)} 
#q(d)
# create a dictionary to store the new key and value in workout_stats dictionary
updated_workStatus ={}

# for loop for calculate the total workout minutes for each friend and
# add the key : vlaue to the variable updated_workStatus
for key, value in workout_stats.items():
   total_workout = sum(value)
   updated_workStatus[key+"_total"] = total_workout

#update dictionary
workout_stats.update(updated_workStatus)

#print updtaed dictionary
print(f"After calculate the total workout minutes for each friend , updated dictionary :\n {workout_stats} \n\n")    


#q(e):2d list= Each row should represent a friend, and each column should represent one activity (yoga, running, or weightlifting)
#declare the list
workout_list=[]

# add to the list
for value in workout_stats.values():
    if isinstance(value, tuple):  # Ensure the value is a tuple so we don't add total minites 
        workout_list.append(value)

#print the list
#print(workout_list)

#q(f)
#	Extract and print the minutes for yoga and running for all friends
# Extract the data and store in to new list                   
yoga_running = [row[:2] for row in workout_list]
#print the list
print("The minutes for yoga and running for all friends.")
for minutes in yoga_running:
    print(minutes)

print()    
# Extract and print the minutes for weightlifting for the last two friends
# Extract the data nad store in to new list
weightlifting = [row[2] for row in workout_list[-2:]]
print("The minutes for weightlifting for the last two friends")
for minutes in weightlifting:
    print(minutes) 

print()


#q(g)
#check the total workout minutes are greater than or equal to 120.and print appropriate message
for key,value in workout_stats.items():
    if isinstance(value, int) :  # Ensure the value is an int 
        if value >= 120:
            print(f"Great job staying active, {key.split('_')[0]} \n")

            

#q(h)a feature to allow the user to input a friend’s name and then print appropriate message according to it
friend_name = input("Enter your friend name: ")
if friend_name in workout_stats:
    print(f"{friend_name} workout minutes for (yoga, running, and weightlifting) activity: {workout_stats[friend_name]} ")
    print(f"And {friend_name} total workout minutes {workout_stats[friend_name + '_total']} \n")
else:
    print(f"Friend {friend_name} not found in the records\n")




#q(i)
# print the  lowest  and highest total workout minutes
total_min= {}
for key,value in workout_stats.items():
    if isinstance(value, int):  # Ensure the value is a int 
        total_min[key]= value


#find the max and min
friend_max = max(total_min,key=total_min.get)
friend_min = min(total_min,key=total_min.get)

# print appropriate message
print(f"Friend with the highest total workout minutes: {friend_max} with {workout_stats[friend_max]} minutes.")
print(f"Friend with the lowest total workout minutes: {friend_min} with {workout_stats[friend_min]} minutes.")



```
